```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
# Upload original image
uploaded = files.upload()
img_path = next(iter(uploaded))
img = cv2.imread(img_path)

# Auto contrast/brightness
def auto_brightness_contrast(image, clip_hist_percent=25):
 gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
 hist = cv2.calcHist([gray],[0],None,[256],[0,256]).flatten()
 acc = np.cumsum(hist)
 clip = clip_hist_percent * (acc[-1] / 100.0) / 2.0
 min_gray = np.searchsorted(acc, clip)
 max_gray = np.searchsorted(acc, acc[-1] - clip)
 alpha = 255 / (max_gray - min_gray + 1e-5)
 beta = -min_gray * alpha
 return cv2.convertScaleAbs(image, alpha=alpha, beta=beta)

enhanced = auto_brightness_contrast(img)
resized = cv2.resize(enhanced, (1080, 1080))
# Show before and after
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
ax[0].set_title("Original")
ax[1].imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
ax[1].set_title("Enhanced & Resized")
for a in ax: a.axis('off')
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
from google.colab import files
import matplotlib.pyplot as plt
from skimage.metrics import structural_similarity as ssim
# Upload two images
uploaded = files.upload()
```

```python
paths = list(uploaded.keys())
# Read and resize
img1 = cv2.imread(paths[0])
img2 = cv2.imread(paths[1])
img1 = cv2.resize(img1, (500, 500))
img2 = cv2.resize(img2, (500, 500))
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
# MSE and SSIM
mse = np.mean((gray1 - gray2) ** 2)
ssim_score, _ = ssim(gray1, gray2, full=True)
print(f"MSE: {mse:.2f}")
print(f"SSIM: {ssim_score:.4f}")
# Show both images
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
ax[0].set_title("Image 1")
ax[1].imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
ax[1].set_title("Image 2")
for a in ax: a.axis('off')
plt.show()
```

```python
# Step 1: Install OpenCV (only needed in Colab)
!pip install opencv-python-headless --quiet
# Step 2: Import required libraries
import cv2
import numpy as np
from google.colab import files
from matplotlib import pyplot as plt
# Step 3: Upload an image
print("Upload an image to detect red objects...")
uploaded = files.upload()
input_image_path = list(uploaded.keys())[0]
# Step 4: Read the image
image = cv2.imread(input_image_path)
if image is None:
 print("Error: Couldn't read the image.")
else:
# Step 5: Convert BGR image to HSV color space
 hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
# Step 6: Define range for red color in HSV
# Red appears in two ranges in HSV (around 0° and 180°)
 lower_red1 = np.array([0, 120, 70])
 upper_red1 = np.array([10, 255, 255])
 lower_red2 = np.array([170, 120, 70])
 upper_red2 = np.array([180, 255, 255])
# Step 7: Create masks for red color
mask1 = cv2.inRange(hsv_image, lower_red1, upper_red1)
mask2 = cv2.inRange(hsv_image, lower_red2, upper_red2)
red_mask = mask1 + mask2
# Step 8: Apply the mask to the original image
red_detected = cv2.bitwise_and(image, image, mask=red_mask)
# Step 9: Show original and result side by side
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(red_detected, cv2.COLOR_BGR2RGB))
plt.title("Red Objects Detected")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
from google.colab import files
import cv2
import numpy as np
from IPython.display import display
from PIL import Image

print("Upload original image:")
uploaded = files.upload()
original_path = list(uploaded.keys())[0]

print("Upload watermark (transparent PNG):")
uploaded2 = files.upload()
watermark_path = list(uploaded2.keys())[0]

def add_watermark(original_path, watermark_path, output_path="watermarked.jpg", transparency=0.6):
    image = cv2.imread(original_path)
    watermark = cv2.imread(watermark_path, cv2.IMREAD_UNCHANGED)

    if image is None or watermark is None:
        print("Error: Could not load images.")
        return

    img_h, img_w = image.shape[:2]

    # Resize watermark if it's larger than the image
    wm_h, wm_w = watermark.shape[:2]
    if wm_h > img_h or wm_w > img_w:
        scale = min(img_h / wm_h, img_w / wm_w) * 0.5
        watermark = cv2.resize(watermark, (0, 0), fx=scale, fy=scale)
        wm_h, wm_w = watermark.shape[:2]

    # Calculate coordinates (bottom right with 10px padding)
    x = max(0, img_w - wm_w - 10)
    y = max(0, img_h - wm_h - 10)

    roi = image[y:y+wm_h, x:x+wm_w]

    # Handle alpha channel for transparency
    if watermark.shape[2] == 4:
        alpha_mask = watermark[:, :, 3] / 255.0
        watermark_rgb = watermark[:, :, :3]
    else:
        alpha_mask = np.ones((wm_h, wm_w))
        watermark_rgb = watermark

    for c in range(3):
        roi[:, :, c] = (
            transparency * alpha_mask * watermark_rgb[:, :, c] +
            (1 - transparency * alpha_mask) * roi[:, :, c]
        )

    image[y:y+wm_h, x:x+wm_w] = roi
    cv2.imwrite(output_path, image)
    print(f"Saved: {output_path}")
    display(Image.open(output_path))

add_watermark(original_path, watermark_path)
```

```python
from google.colab import files
import cv2
import numpy as np
from IPython.display import display
from PIL import Image
import matplotlib.pyplot as plt

print("Upload scanned document image:")
uploaded = files.upload()
input_path = list(uploaded.keys())[0]

def preprocess_for_ocr(input_path, output_path="ocr_ready.jpg"):
    image = cv2.imread(input_path)
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    denoised = cv2.GaussianBlur(gray, (5, 5), 0)
    _, threshold = cv2.threshold(denoised, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

    cv2.imwrite(output_path, threshold)

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    axes[0].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)); axes[0].set_title("Original"); axes[0].axis("off")
    axes[1].imshow(denoised, cmap="gray");                  axes[1].set_title("Denoised"); axes[1].axis("off")
    axes[2].imshow(threshold, cmap="gray");                 axes[2].set_title("OCR Ready"); axes[2].axis("off")
    plt.tight_layout()
    plt.show()
    print(f"Saved: {output_path}")

preprocess_for_ocr(input_path)
```

```python
from google.colab import files
import cv2
import os
import matplotlib.pyplot as plt

print("Upload all images to convert to grayscale:")
uploaded = files.upload()

def batch_convert_to_grayscale(uploaded_files, output_folder="output_bw"):
    os.makedirs(output_folder, exist_ok=True)
    converted = []

    for filename, data in uploaded_files.items():
        with open(filename, "wb") as f:
            f.write(data)

        image = cv2.imread(filename)
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

        name, ext = os.path.splitext(filename)
        output_path = os.path.join(output_folder, f"{name}_bw{ext}")
        cv2.imwrite(output_path, gray)
        converted.append((filename, output_path, image, gray))
        print(f"{filename}  →  {output_path}")

    fig, axes = plt.subplots(len(converted), 2, figsize=(10, 4 * len(converted)))
    if len(converted) == 1:
        axes = [axes]
    for i, (orig_name, _, orig, gray) in enumerate(converted):
        axes[i][0].imshow(cv2.cvtColor(orig, cv2.COLOR_BGR2RGB)); axes[i][0].set_title(f"Original: {orig_name}"); axes[i][0].axis("off")
        axes[i][1].imshow(gray, cmap="gray"); axes[i][1].set_title("Grayscale (_bw)"); axes[i][1].axis("off")
    plt.tight_layout()
    plt.show()
    print(f"\nDone! {len(converted)} image(s) converted.")

batch_convert_to_grayscale(uploaded)
```

```python
from google.colab import files
import cv2
import matplotlib.pyplot as plt

print("Upload a noisy scanned image:")
uploaded = files.upload()
input_path = list(uploaded.keys())[0]

def remove_noise_and_enhance(input_path, output_path="clean_scan.jpg", method="gaussian"):
    image = cv2.imread(input_path)
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    if method == "gaussian":
        filtered = cv2.GaussianBlur(gray, (5, 5), 0)
    elif method == "median":
        filtered = cv2.medianBlur(gray, 5)

    threshold = cv2.adaptiveThreshold(
        filtered, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY, 11, 2
    )

    cv2.imwrite(output_path, threshold)

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    axes[0].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)); axes[0].set_title("Original (Noisy)"); axes[0].axis("off")
    axes[1].imshow(filtered, cmap="gray");                  axes[1].set_title(f"After {method.capitalize()} Filter"); axes[1].axis("off")
    axes[2].imshow(threshold, cmap="gray");                 axes[2].set_title("Final (Thresholded)"); axes[2].axis("off")
    plt.tight_layout()
    plt.show()
    print(f"Saved: {output_path}")

remove_noise_and_enhance(input_path, method="gaussian")  # Change to "median" if needed
```
